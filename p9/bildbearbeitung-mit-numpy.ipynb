### 9.2.4 Kantenerkennung
Die Kantenerkennung war die komplizierteste Funktion der Bildbearbeitung in Aufgabe $7$.
<center>

![Kantenerkennung](./figures/kantenerkennung.png)<br>
Abbildung 9.2.4.1.: Kantenerkennung durch gewichtete Mittelwertbildung über 9 Bildpunkte
</center>

Wie bei der Glättung wurden alle Pixel auf den Mittelwert der Graustufenwerte des Pixels selbst und seiner 8 umgebenden Pixel gesetzt. Zuvor wurden die Graustufenwerte aber mit den Werten einer Gewichtungsmatrix multipliziert. Da das Ergebnis negativ sein konnte, wurde für jeden Graustufenwert der Betrag (`abs()`) des Ergebnisses berechnet. Für ein Pixel an der gegebenen Position `y` und `x` sah das etwa so aus:
```Python
pixel = 0
# Anfang (1)
for dy in [-1, 0, 1]:
    for dx in [-1, 0, 1]:
        pixel += image['pixels'][y+dy][x+dx] * weights[w_y][w_x]
# Ende (1)
newValue = abs(int(pixel/9))
```

#### 9.2.4.1 Faltung
In dem Code ist ein Block mit "Anfang (1)" bis "Ende (1)" markiert. In diesem Block wird für ein Pixel die Summe des Grauwerts des Pixels und seiner 8 umgebenden Pixel berechnet, die zuvor mit dem Gewichtungsfaktor aus der Gewichtungsmatrix multipliziert wurde.<br>
Diese Funktion, angewandt auf die Graustufenwerte aller Pixel (außer den äußersten Pixel des Bildes (Rahmen), auf die die Matrix nicht angewandt werden kann) wird *Faltung* genannt. Diese Funktion und diesen Begriff werden Module höherer Semester noch einmal aufgreifen.

Für die *Faltung* steht eine eigene Funktion zur Verfügung:
```Python
from scipy.signal import convolve2d
...

# Gewichtungsmatrix
weights = np.array([
        [0, -1, 0], 
        [-1, 4 ,-1], 
        [0, -1, 0]], 
        dtype=np.float32)
    
# Zweidimensionales numpy.ndarray mit den Graustufen, konvertiert von int-Werten in 32-Bit-Float Werte
img_f = image['pixels'].astype(np.float32)

# Nur Faltung - KEIN Mittelwert (!!!!)
convolved = convolve2d(img_f, weights, mode='same', boundary='fill', fillvalue=0)
```
* `img_f` ist das zu bearbeitende Array mit den Graustufenwerten im `np.float32`-Datenformat.
* `weights`ist die zuvor definierte Gewichtungsmarix mit den Gewichtungsfaktoren den Pixel im `np.float32`-Datenformat.
* `mode='same'` bedeutet, dass das zurückgegeben Array genau so gross ist, wie das zu bearbeitende Array. Es gibt mehrere Einstellmöglichkeiten, z.B. dass ein verkleinertes Array zurückgegeben wird, das nur die Graustufenwerte enthält, die wirklich durch die Faltung bearbeitet werden konnten (kein unbearbeiteter Rand).
* `boundary='fill'` stellt ein, was mit dem unbearbeiteten Rand geschehen soll. In diesem Fall soll er automatisch mit einem Wert gefüllt werden.
* `fillvalue=0` stellt den Wert ein, mit dem die Füllung erfolgen soll. Die Voreinstellung ist `0`, also schwarz. Da der Rand laut Aufgabenstellung mit "Schwarz" gefüllt werden soll, kann die Voreinstellung verwendet und braucht nicht unbedingt angegeben werden.



* Die Funktion `convolved(...)` verwendet für die Berechnung den gegebenen Datentyp der Graustufenwerte im Array, der beim Einlesen auf<br>
`dtype=np.uint8` gesetzt wurde. Dieser Datentyp würde bei den anstehenden Berechnungen über- ($grauwert > 255$) oder unterlaufen ($grauwert < 0$) und ist damit ungeeignet. Daher werden die Werte im Array durch solche in einem 32-Bit-Float-Format ersetzt. Ein 64-Bit-Float-Wert wäre ebenso möglich.<br>
Im Ergebnis-Array `convolved` liegen die Graustufen dann ebenfalls im 32-Bit-Float-Format vor.

Um die Kanten in einem Bild erkennbar zu machen, reicht diese Operation aus.<br>
Allerdings werden viele Graustufen den erforderlichen Wertebereich für "P2"-PGM Graustufenbilder von `0` bis `255` sowohl unter- als auch überschreiten. Außerdem ist der Datentyp `np.float32` nicht für das Schreiben in eine Bilddatei geeignet. Wenn Sie das Bild in diesem Zustand zur Ausgabe zurück geben wollen, müssen Sie zuvor noch folgende Funktion ausführen:
```Python
result['pixels'] = np.clip(convolved, 0, 255).astype(np.uint8)
```
* Die Funktion `np.clip(convolved, 0, 255)` begrenzt die Werte im übergebenen Array auf den unteren Grenzwert `0` und den oberen Grenzwert `255`. Werte die darüber und darunter liegen, werden durch den Grenzwert `0` oder `255` ersetzt. Das Ergebnis ist wieder ein Array mit Werten mit dem Datentyp `np.float32`.
* Die auf das zurück gegebene Array angewandte Funktion `.astype(np.uint8)` konvertiert alle Werte im Array in den Datentyp `np.uint8`, dem vorzeichenlosen 8-Bit-Ganzzahltyp, der für das Schreiben auf in eine Datei benötigt wird.

Das dabei herauskommende Bild können Sie sich schon einmal ansehen, bevor die Funktion fortgeführt wird.

#### 9.2.4.2 Mittelwertbildung
Nach der Faltung entspricht das Ergebnisbild noch nicht ganz dem Ergebnisbild der Kantenerkennung aus Aufgabe 7.

Um die Bearbeitung fortzusetzen, löschen Sie zunächst die Zeile 
```Python
result['pixels'] = np.clip(convolved, 0, 255).astype(np.uint8)
```
Danach geht die Bearbeitung mit folgenden Schritten weiter:
* Um den Mittelwert zu bestimmen, müssen die Graustufenwerte aller Pixel in der Ergebnismatrix durch $9$ dividiert werden.
* Um negative Werte zu vermeiden, müssen diese durch Betragsbildung positiv werden.
* Für die Rückgabe im korrekten Format müssen auch hier die Graustufenwerte in das vorzeichenlosen 8-Bit-Format konvertiert werden.

Die Zeile, die alle Aufgaben auf einmal erledigt, sieht so aus:
```Python
result['pixels'] = np.abs(np.trunc(convolved / 9)).astype(np.uint8)
```
* `convolved / 9` dividiert zunächst alle Werte der gefalteten Werte im Ergebnis-Array durch `9`, um den Mittelwert zu bestimmen. Das Datenformat ist hier noch `np.float32`.
* np.trunc(..) führt auf alle Werte im Array anschließend die `truncate`-Funktion aus, die Bruchanteile aller Float-32-Werte "abschneidet". Das Ergebnis sind ganze Zahlen, aber immer noch im `np.float32`-Format. Negative Werte sind immer noch negativ.
*  np.abs(...) führt als nächstes die Betragsfunktion auf alle Graustufenwerte im Array aus, so dass das Array nur noch positive Werte enthält. Immer noch im `np.float32`-Format.
* .astype(np.uint8) konvertiert die Werte im Array abschließend in das für die Rückgabe erforderliche vorzeichenlose 8-Bit-Ganzzahlformat.

**<u>Wichtig</u>**<br>
