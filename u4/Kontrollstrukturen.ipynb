## Erweiterung der `if`-Anweisung durch `elif` und `else`

### Logische Operatoren

Die folgenden logischen Operatoren können in `if`-Anweisungen verwendet werden:

* Gleichheit (`==`)
* Ungleichheit (`!=`)
* Größer als (`>`)
* Kleiner als (`<`)
* Größer oder gleich (`>=`)
* Kleiner oder gleich (`<=`):

Experimantieren Sie mit dem nachfolgenden Codeblock um sich mit der Verwendung der logischen Operatoren vertraut zu machen.

```python
a=5
b=5

print("Werte von a und b:")
print("a =", a)
print("b =", b)

if a == b:
    print("a ist gleich b")

if a != b:
    print("a ist nicht gleich b")

if a > b:
    print("a ist größer als b")

if a < b:
    print("a ist kleiner als b")

if a >= b:
    print("a ist größer oder gleich b")

if a <= b:
    print("a ist kleiner oder gleich b")
```

Die Verwendung von `elif` (else if) und `else` anstelle von mehreren separaten `if`-Anweisungen hat Vorteile:

1. Verbesserte Lesbarkeit:
   - *Klarheit der Struktur*: Die Verwendung von `elif` und `else` schafft eine klare hierarchische Struktur. Es wird deutlicher, dass die verschiedenen Bedingungen in einer Abhängigkeit zueinander stehen.
   - *Weniger Codezeilen*: Durch die Vermeidung redundanter `if`-Bedingungen wird der Code kürzer und übersichtlicher.
2. Eindeutige Entscheidung:
   - *Einmalige Ausführung*: Mit `elif` und `else` wird genau eine der Bedingungen ausgeführt. Wenn ein `if`-Zweig wahr ist, werden die nachfolgenden Bedingungen nicht mehr überprüft. Das verhindert unnötige Berechnungen.
   - *Reduzierte Komplexität*: Bei mehreren `if`-Anweisungen besteht die Gefahr, dass *mehrere Bedingungen* wahr sind, und das Programm eine unerwartete oder unerwünschte Ausgabe produziert.
3. Bessere Fehlervermeidung:
   - Reduzierte Wahrscheinlichkeit für Fehler: Bei vielen `if`-Anweisungen kann leicht das Risiko steigen, dass logische Fehler auftreten, etwa durch das Vergessen eines `break` oder das Einfügen falscher Bedingungen. Die Verwendung einer klaren `if`, `elif`, `else`-Struktur verringert dieses Risiko.
3. Effizienz:
   - *Potentiell schnellerer Code*: Der Code mit `elif` und `else` kann effizienter sein, da er nicht alle Bedingungen prüfen muss, wenn eine bereits erfüllt ist. Wenn Sie viele `if`-Anweisungen verwenden, wird jede Bedingung unabhängig voneinander geprüft, was Zeit und Ressourcen beansprucht.

Die folgenden $2$ Code-Zellen sollen dies verdeutlichen:

```python
note = 85

if note >= 90:
    print("Sehr gut")
if note >= 80:
    print("Gut")  # Auch wenn note >= 90, wird diese Bedingung ausgewertet, was zu Verwirrung führen kann
if note < 80:
    print("Verbesserung nötig")


note = 90

if note >= 90:
    print("Sehr gut")
if note >= 80:
    print("Gut")  # Auch wenn note >= 90, wird diese Bedingung ausgewertet, was zu Verwirrung führen kann
if note < 80:
    print("Verbesserung nötig")
```

besser so:

```python
note = 85

if note >= 90:
    print("Sehr gut")
elif note >= 80:
    print("Gut")
else:
    print("Verbesserung nötig")
```

#### Logische Operatoren bei Zeichenketten (Strings)

Der Gleichheitsoperator (`==`) und Ungleichheitsoperator (`!=`) verhalten sich *wie erwartet*:

```python
str1 = "Hallo"
str2 = "Hallo"
str3 = "hallo"

print(str1 == str2)  # Ausgabe: True (Inhalt ist gleich)
print(str1 == str3)  # Ausgabe: False (Unterschiedliche Großschreibung)
```

```python
print(str1 != str2)  # Ausgabe: False (sind gleich)
print(str1 != str3)  # Ausgabe: True (sind unterschiedlich)
```

Die Vergleichsoperatoren (größer-als-Operator (`>`) und kleiner-als-Operator (`<`) (sowie die entsprechenden Operatoren `>=`und `<=`)) funktionieren bei Zeichenketten ähnlich wie bei numerischen Werten, jedoch mit einem [lexikografischen](https://de.wikipedia.org/wiki/Lexikographische_Ordnung) Vergleich, bei dem die Zeichen entsprechend ihrer [ASCII](https://de.wikipedia.org/wiki/American_Standard_Code_for_Information_Interchange)-Werte (oder [Unicode](https://de.wikipedia.org/wiki/Unicode)-Werte) verglichen werden. Diese Eigenschaften sind nützlich, wenn Sie mit Textdaten arbeiten und Bedingungen auf der Grundlage des Inhalts oder der Reihenfolge von Strings erstellen möchten.

Großbuchstaben kommen dabei vor Kleinbuchstaben.

```python
str4 = "Apfel"
str5 = "Birne"

print(str4 > str5)  # Ausgabe: False (Da 'A' < 'B')

print("1.", "Zebra" > "Apfel")  # Ausgabe: True ('Z' ist größer als 'A')

print("2.", str4 < str5)  # Ausgabe: True (Da 'A' < 'B')

print("3.", "Hallo" >= "Hallo")  # Ausgabe: True

print("4.", "Zebra" >= "Apfel")   # Ausgabe: True

print("5.", "Apfel" <= "Birne")  # Ausgabe: True

print("6.", "Herr" <= "Haus")   # Ausgabe: ?

print("7.", "haus" > "Haus")   # Ausgabe: ?

print("8.", "1" > "Eins")

print("9.", str(1) > "E")   

print(str(1))
```

## Vergleich zwischen `==` und `is` in Python

- `==` vergleicht den Wert von zwei Objekten. Wenn die Werte gleich sind, gibt der Operator `True` zurück, andernfalls `False`
- `is` vergleicht die Identität zweier Objekte, d.h. es prüft, ob zwei Variablen auf dasselbe Objekt im Speicher verweisen. Wenn sie dasselbe Objekt sind, gibt der Operator `True` zurück, andernfalls `False`.

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)  # Ausgabe: True (da die Listen die gleichen Werte haben)
```

```python
x = "5"
y = 5

print(x == y)  # Ausgabe: False (da die Typen unterschiedlich sind)
```

```python
a = [1, 2, 3]
b = a  # b verweist auf dasselbe Objekt wie a

print(a is b)  # Ausgabe: True (da beide auf dasselbe Objekt zeigen)

c = [1, 2, 3]  # c ist ein neues Objekt

print(a is c)  # Ausgabe: False (da c ein neues Objekt ist)
```

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a == c)  # Ausgabe: True (gleiche Werte)
print(a is c)  # Ausgabe: False (verschiedene Objekte)
print(a is b)  # Ausgabe: True (dasselbe Objekt)
```

Mit `copy()`können sie erzwingen, dass auch wirklich neuer Speicher für die Variable `b`angelegt wird.

```python
a = [1, 2, 3]
b= a.copy()

print(a is b)
```

## Fehlerbehandlung mit `try` und `except`

In Python wird mit `try` und `except` die Fehlerbehandlung implementiert, um Programmausnahmen (Exceptions) abzufangen und eine kontrollierte Reaktion auf Fehler zu ermöglichen. Dies trägt dazu bei, dass ein Programm nicht abrupt abbricht, wenn ein Fehler auftritt.

1. `try`-Block:
  - Der Code, der möglicherweise einen Fehler verursachen könnte, wird in den `try`-Block geschrieben. Wenn ein Fehler innerhalb dieses Blocks auftritt, wird der normale Programmfluss gestoppt und zum except-Block gesprungen.
2. `except`-Block:
   - Hier wird der Code platziert, der ausgeführt werden soll, wenn ein Fehler auftritt. Man kann spezifische Fehler angeben (z. B. `ValueError`, `ZeroDivisionError`) oder einen allgemeinen Block für alle Ausnahmen verwenden.

Der folgende Codeabschnitt soll das verdeutlichen indem er eine Division durch den Wert $0$ abfängt.

```python
# devide 10 by the number given by the user
try:
    zahl = int(input("Bitte eine Zahl eingeben: "))
    result = 10 / zahl
    print(f"Das Ergebnis ist: {result}")
except ValueError:
    print("Das war keine gültige Zahl!")
except ZeroDivisionError:
    print("Division durch Null ist nicht erlaubt!")

print("Programm läuft weiter...")
```

```python
try:
    # Code, der einen Fehler verursachen könnte
    dateiname = "non_existent_file.txt"
    with open(dateiname, 'r') as file:
        data = file.read()
except Exception as e:
    print(f"Ein Fehler ist aufgetreten: {e}")  # Gibt den Fehler aus

print("Programm läuft weiter...")
```

Zusätzlich zu `try` und `except` gibt es auch die optionalen Blöcke `else` und `finally`:

```python
try:
    zahl = int(input("Bitte eine Zahl eingeben: "))
    result = 10 / zahl
except ZeroDivisionError:
    print("Division durch Null ist nicht erlaubt!")
else:
    print(f"Das Ergebnis ist: {result}")
```
