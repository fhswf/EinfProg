# Objektorientierte Programmierung (OOP)

Die Einführung in die Programmierung basierte bisher auf funktionsorientierter Programmierung.<br>
Das Vorgehen bei der Lösung von Aufgaben umfasste folgende Schritte:
* Aufgaben in Teilaufgaben / Teilprobleme zu zerlegen,
* Algorithmen zur Bearbeitung / Lösung der Teilaufgaben / Teilprobleme zu entwickeln und
* die Algorithmen in der Programmiersprache Python auszudrücken.

Nach diesem Vorgehen werden Aufgabe als Ganzes, in "einem Block" gelöst, auch wenn Programmteile zur zur bessern Übersicht in Funktionen ausgelagert werden.<br> 
Vergleichen kann man das Ganze mit der Konstruktion eines heutigen Smartphones, das eine Platine hat, die als Ganzes funktioniert - oder eben auch nicht.

**Objektorientierte Programmierung erfordert eine ganz andere Denkweise!**

Auch hier müssen die Aufgaben zunächst in Teilaufgaben zerlegt werden.

Die Lösungen der Teilaufgaben werden dann in einzelne Klassen abgebildet. Die Objekte dieser Klassen enthalten alle Funktionen und erforderlichen Daten (Objektvariablen), die zudem gegen Manipulation von außen gekapselt sind bzw. sein müssen.<br>
Diese Objekte lösen je eine **einzige (!!!)** klar umrissene Teilaufgabe (**Zuständigkeit eines Objekts** genannt).<br>
Diese Beschränkung auf exakt eine Zuständigkeit wird **Single responsibility Priciple** genannt. Ziel ist es, statt schwer überschaubaren, funktionsorientierten Spaghetticode (Motto: "Geht doch!"), Code zu entwickeln, der
* gut wiederverwendbar,
* gut erweiterbar und
* gut wartbar ist.

Die Gesamtaufgabe wird gelöst, indem die einzelnen, erforderlichen Objekte so miteinander kombiniert werden, dass das Zusammenspiel die Lösung ergibt.

Man kann sich die Objekte einer Anwendung wie abgeschlossene Funktionseinheiten vorstellen, die ähnlich wie Karosserie, Motor, Getriebe, Achsen, etc. im Auto jede für sich ihre Funktion erfüllen und erst zusammengebaut und im Zusammenspiel ein funktionsfähiges Auto ergeben.

## Erstes Beispiel: Erstellen einer Klasse

### Definition der Klasse

Wir erstellen im folgenden Code Abschnitt eine einfache Klasse `Hund` mit den Eigenschaften (Attributen) `name` und `rassse`, sowie der Methode `bellen()`.

````python
class Hund:
    def __init__(self, name, rasse):
        self.name = name
        self.rasse = rasse

    def bellen(self):
        return f"{self.name} sagt: Wuff!"
````

### Instanziierung / Initialisierung der Klasse:

Nun können wir eine *Instanz* der Klasse (also eine Variable vom Typ `Hund`) anlegen.

````python
mein_hund = Hund("Bello", "Labrador")

print(mein_hund.bellen())  # Ausgabe: Bello sagt: Wuff!

type(mein_hund)  # Ausgabe: <class '__main__.Hund'>
````

### `self`, `__init__()`,  `__new__()` und `__del__()`

In der Definition der Klasse `Hund` mag "*der Bezeichner*" `self` auffallen, sowie die Funktion `__init__()`.

- `self` stellt eine Referenz auf die aktuelle Instanz der Klasse dar. Wenn Sie ein Objekt einer Klasse erstellen, können Sie mit `self` auf dessen Daten und Funktionen zugreifen.
- `self` muss immer als erstes Argument in den Methoden der Klasse stehen. Dies zeigt an, dass die Methode einer Instanz der Klasse aufgerufen wird.
- Mit `self` können Instanzattribute (Variablen) definieren werden und darauf zugegriffen werden. 
- Die Nutzung des Bezeichners `self` ist (nur) eine Konvention. `self` ist kein reserviertes Wort; Sie könnten es theoretisch auch anders benennen. Es ist jedoch ratsam, bei `self` zu bleiben, um den Code verständlich zu halten.

- Der Konstruktor `__init__()` initialisiert die Attribute `name` und `rassse` und verwendet `self`, um sicherzustellen, dass diese Attribute zur Instanz gehören.
- Der Konstruktor ist eine spezielle Methode in Python, die aufgerufen wird, wenn ein neues Objekt der Klasse erstellt wird. Die Methode `__init__()' initiiert die Eigenschaften des Objekts und wird verwendet, um ihre Attribute festzulegen.
- Die `__init__()`-Methode wird automatisch aufgerufen, nachdem das Objekt erfolgreich erstellt wurde.
  - In Beispiel oben wird `__init__()` verwendet, um die Attribute `name` und `rassse` des `mein_hund`-Objekts festzulegen, sobald es instanziiert wird.
- Es gibt neben dem Konstruktor `__init()` noch die Funktion `__new__` die  für die Erstellung des Objekts zuständig ist (während `__init__()` für die Initialisierung der Attribute des Objekts verantwortlich ist. In den meisten Fällen werden Sie nur `__init__()` verwenden, es sei denn, Sie müssen spezifische Kontrolle über den Instanziierungsprozess haben, wobei Sie `__new__()` benötigen könnten.

Die folgende Tabelle fasst die Unterschiede zwischen `__new__()` und `__init__()` zusammen:

Aspekt        | `__new__()`                            | `__init__()`
--------------|----------------------------------------|-----------------------------------------------
Aufruf        | Vor der Erstellung des Objekts         | Nach der Erstellung des Objekts
Zweck         | Erstellt das Objekt und gibt es zurück | Initialisiert das Objekt
Notwendigkeit | Wird selten benötigt.                  | Häufig verwendet, z.B. um Attribute zu setzen

Im folgenden Beispiel wird `__new__()` verwendet um eine Nachricht auszugeben, bevor das neue Objekt erstellt wird. Hier ist zu beachten, dass der Aufruf von `super()` in `__new__()` notwendig ist, um tatsächlich das neue Objekt zu erstellen (dazu später mehr).

Analog zum Konstruktor gbt es noch den Destruktor `__del__()`, der in dem Moment aufgerufen wird, kurz bevor ein Objekt zerstört wird. Der Aufruf des Destruktors macht z.B. Sinn, um geöffnete Dateien zu schließen bevor die entsprechende Instanz der Klasse zerstört wird.

````python
class Auto:
    def __new__(cls, *args, **kwargs):
        print("Ein neues Objekt wird erstellt.")
        # Objekt erstellen
        return super(Auto, cls).__new__(cls)

    def __init__(self, marke, modell):
        self.marke = marke
        self.modell = modell
        print(f"Auto fertig initialisiert: {self.marke} {self.modell}")

    def __del__(self):
        print(f"Auto gelöscht: {self.marke} {self.modell}")

# Objekt erstellen
mein_auto = Auto("Volkswagen", "Golf")

# instanz löschen
del mein_auto
````

### Klassenmethoden

Im vorherigen Beispiel haben wir statt `self`den Bezeichner `cls` verwendet. Dieser deutet auf eine **Klassenmethode** hin. 

Klassenmethoden sind Methoden, die auf die *Klasse selbst* und nicht auf eine *Instanz der Klasse* operieren. Sie werden verwendet, um auf Klassendaten oder zur Erstellung von Instanzen über verschiedene Konstruktor-Logiken hinweg zuzugreifen. Klassenmethoden werden mit dem Decorator `@classmethod` gekennzeichnet.

- Das erste Argument einer Klassenmethode ist immer `cls`, das auf die Klasse selbst verweist (nicht auf eine Instanz). Hiermit hat man Zugriff auf die Klasse.
- Klassenmethoden sind nützlich, um Funktionalitäten zu implementieren, die für die Klasse insgesamt relevant sind und nicht für einzelne Objekte (Instanzen).

````python
class Fahrzeug:
    @classmethod
    def info(cls):
        return f"Dies ist eine Instanz der Klasse {cls.__name__}."

print(Fahrzeug.info())  # Ausgabe: Dies ist eine Instanz der Klasse Fahrzeug.
````

````python
class Buch:
    allgemeine_kategorie = "Literatur"

    def __init__(self, titel, autor):
        self.titel = titel
        self.autor = autor

    @classmethod
    def aus_kategorie(cls):
        return cls.allgemeine_kategorie

# Aufruf der Klassenmethode
print(Buch.aus_kategorie())  # Ausgabe: Literatur
````

In diesem Beispiel ist `aus_kategorie()` eine Klassenmethode, die die allgemeine Kategorie des Buches zurückgibt.
Sie verwendet `cls`, um auf die Klasse Buch zuzugreifen und damit auf das Klassenattribut `allgemeine_kategorie`.

Klassenmethoden können als *alternative Konstruktoren* dienen, um Instanzen auf verschiedene Weisen zu erstellen. Hier ein Beispiel:

````python
class Auto:
    def __init__(self, marke, modell):
        self.marke = marke
        self.modell = modell

    @classmethod
    def von_string(cls, auto_string):
        marke, modell = auto_string.split('-')
        return cls(marke, modell)  # Erstellung einer Instanz

# Alternative Konstruktion
mein_auto = Auto.von_string("Volkswagen-Golf")
print(mein_auto.marke)  # Ausgabe: Volkswagen
print(mein_auto.modell)  # Ausgabe: Golf
````

Die `math`-Bibliothek in Python, die wir schon am Rande kennen gelernt haben, ist ein gutes Beispiel für die Verwendung von Klassenmethoden. Obwohl die `math`-Bibliothek nicht in Klassen aufgebaut ist, können wir dieselben Konzepte auf Klassen mit ähnlichem Verhalten anwenden. 

````python
class Mathe:
    @classmethod
    def quadrat(cls, zahl):
        """Berechnet das Quadrat einer Zahl."""
        return zahl ** 2

    @classmethod
    def umrechnen(cls, von, zu, faktor):
        """Konvertiert von einer Einheit in eine andere unter Verwendung eines Faktors."""
        return von * faktor / zu

# Nutzung der Klassenmethoden
print(Mathe.quadrat(4))  # Ausgabe: 16
print(Mathe.umrechnen(100, 1, 0.001))  # Ausgabe: 0.1 (z.B. (100) mm in (1) m)
````

Diese Klassenmethoden im Beispiel oben ermöglicht es, die Methode ohne Instanziierung der Klasse aufzurufen.

### Klasseneigenschaften

In Fällen, in denen eine Klasse Attribute oder Eigenschaften hat, bietet eine Klassenmethode Zugriff auf diese. Dies ist besonders nützlich für Fabrikmethoden oder zur Verwendung von Klassendaten.

### Kapselung
Daten (Attribute) und die dazugehörigen Funktionen (Methoden) werden in einer Klasse in einer Einheit gebündelt und der direkten Zugriff auf die Daten von außen wird beschränkt. Dies dient zum
- Schutz der Datenintegrität und zur 
- Reduzierung der Komplexität,
indem Benutzer nur mit definierten Methoden interagieren.

Die folgende Klasse *implementiert ein Bankkonto*. Auf den Kontostand (Variable `__kontostand`) kann nicht direkt zugegriffen werden (nur über die zur Verfügung gestellten Methoden (z.B. `einzahlen()`, `abheben()`,...))

````python
class Bankkonto:
    def __init__(self, kontostand=0):
        self.__kontostand = kontostand  # private Variable

    def einzahlen(self, betrag):
        self.__kontostand += betrag

    def abheben(self, betrag):
        if betrag <= self.__kontostand:
            self.__kontostand -= betrag
        else:
            print("Nicht genügend Guthaben!")

    def anzeigen(self):
        return self.__kontostand
````

````python
mein_konto = Bankkonto()
print(mein_konto.anzeigen())  # Ausgabe: 0
mein_konto.einzahlen(100)
print(mein_konto.anzeigen())  # Ausgabe: 100
mein_konto.abheben(50)
print(mein_konto.anzeigen())  # Ausgabe: 50
mein_konto.abheben(100)  # Ausgabe: Nicht genügend Guthaben!
````

Auf die Variable `__kontostand` der Klasse `Bankkonto` kann nicht von außerhalb der Klasse zugegriffen werden, da diese *privat* ist (dazu später noch mehr).

````python
print(mein_konto.__kontostand)  # Fehler: AttributeError, da __kontostand privat ist
````

#### Verwendung von Unterstrichen in Python

Die doppelten Unterstriche in Python, wie z.B. bei `__init__`, haben spezielle Bedeutungen und dienen hauptsächlich der Namensverknüpfung (Name Mangling) und der Vermeidung von Namenskonflikten. Hier sind die wichtigsten Punkte:

1. Spezielle Methoden (Dunder-Methoden)
   - Die doppelten Unterstriche markieren spezielle Methoden, die oft als „Dunder-Methoden“ (Double Underscore Methods) bezeichnet werden. Beispielhafte Methoden sind:
     - `__init__`: Der Konstruktor, der beim Erstellen eines Objekts aufgerufen wird.
     - `__str__`: Definiert, wie das Objekt als String dargestellt wird.
     - `__repr__`: Gibt eine offizielle String-Repräsentation des Objekts zurück.
   - Diese Methoden haben spezifische Aufgaben und können von Python intern aufgerufen werden. 
   - Die doppelten Unterstriche sind nicht nur eine Konvention, sondern ein wichtiges Merkmal der Python-Programmierung, das die Struktur und Sicherheit von Code verbessert. Sie führen zu einer klareren Trennung zwischen öffentlichen und privaten Attributen und Methoden.

2. Namensverknüpfung (Name Mangling)
   - Bei Attributen, die mit doppelten Unterstrichen definiert sind (z.B. __attribute), wandelt Python den Namen intern um, um Namenskonflikte in abgeleiteten Klassen zu vermeiden. Das bedeutet, dass:
    - __attribute in der Klasse MyClass intern als _MyClass__attribute gespeichert wird.
Dies macht es schwierig, direkt auf diese Attribute von außen zuzugreifen, fördert aber den sicheren Zugriff auf die Attribute innerhalb der Klasse.

3. Kapselung von Attributen
   - Durch die Verwendung von doppelten Unterstrichen wird eine Form von Kapselung erreicht. Das ermöglicht es, interne Implementierungsdetails eines Objekts zu verbergen.

Beispiel
Hier ist ein einfaches Beispiel, das die Verwendung von `__init__` und doppelten Unterstrichen verdeutlicht:

````python
class BeispielKlasse:
    def __init__(self, wert):
        self.__wert = wert  # private Attribut mit Namensverknüpfung
        self._wert = wert   # geschütztes Attribut
        self.wert = wert    # öffentliches Attribut

    def zeige_wert(self):
        return self.__wert

objekt = BeispielKlasse(10)
print(objekt.zeige_wert())  # Ausgabe: 10
````

### Platzierung von Unterstrichen

- Am Anfang eines Bezeichners:

  - `self._attribut`: wird als **protected** betrachtet, d.h. es soll nur innerhalb der Klasse und von abgeleiteten Klassen verwendet werden.
  - `self.__attribut`: wird durch Namensverknüpfung geschützt und ist **privat**“** für die Klasse.

- Am Ende eines Bezeichners:

  - `name_` oder `name__`: Hierbei handelt es sich normalerweise um die Konvention, um Reservierungskonflikte mit Schlüsselwörtern oder um spezielle Funktionen zu vermeiden. Der Unterstrich am Ende wird verwendet, um Konflikte mit eingebaute Namen zu vermeiden.


````python
print(objekt._wert)  # Dies sollte man nicht tun, aber es funktioniert
print(objekt.wert)   # Öffentliches Attribut, kein Problem
````

````python
print(objekt.__wert)  # Dies führt zu einem Fehler
````

### Vererbung

Bei der Vererbung können untergeordnete Klassen die übergeordneten Klassenmethoden überschreiben oder erweitern und so anpassbare Funktionalität bereitstellen.

Beispiel 1:

````python
class Hund:
    def __init__(self, name, rasse):
        self.name = name
        self.rasse = rasse

    def bellen(self):
        return f"{self.name} sagt: Wuff!"

class Mops(Hund):
    def bellen(self):
        return f"{self.name} sagt: Wuff! (aber anders)"
````

Beispiel 2:

````python
class Tier:
    def sprechen(self):
        return "Tiergeräusch"

class Hund(Tier):  # Hund erbt von Tier
    def sprechen(self):
        return "Wuff!"
````

````python
class Tier:
    @classmethod
    def geräusch(cls):
        return "Ein Tiergeräusch"

class Hund(Tier):
    @classmethod
    def geräusch(cls):
        return "Wuff!"

print(Hund.geräusch())  # Ausgabe: Wuff!
````

### Erben von mehreren Klassen

leichzeitige Vererbung (auch Multiple Vererbung genannt) ist ein Feature in Python, bei dem eine Klasse von mehr als einer Oberklasse erbt. Dies kann nützlich sein, um Funktionen und Eigenschaften aus mehreren Basisklassen in einer Unterklasse zu kombinieren. Im folgenden Code-Abschnitt erbt die Klasse `Auto` von zwei Klassen (`Motorfahrzeug` und `Kraftstoff`).

````python
# Basis-Klasse 1
class Motorfahrzeug:
    def __init__(self, marke, modell):
        self.marke = marke
        self.modell = modell
    
    def fahrzeug_info(self):
        return f"{self.marke} {self.modell}"

# Basis-Klasse 2
class Kraftstoff:
    def __init__(self, kraftstoff_typ):
        self.kraftstoff_typ = kraftstoff_typ
    
    def kraftstoff_info(self):
        return f"Mein Fahrzeug läuft mit {self.kraftstoff_typ}."

# Unterklasse, die von beiden Basisklassen erbt
class Auto(Motorfahrzeug, Kraftstoff):
    def __init__(self, marke, modell, kraftstoff_typ):
        Motorfahrzeug.__init__(self, marke, modell)
        Kraftstoff.__init__(self, kraftstoff_typ)

    def auto_info(self):
        return f"{self.fahrzeug_info()}. {self.kraftstoff_info()}"

# Verwendung
mein_auto = Auto("Volkswagen", "Golf", "Benzin")
print(mein_auto.auto_info())
````

### Polymorphismus

Polymorphismus ermöglicht es, dass verschiedene Klassen Methoden mit dem gleichen Namen implementieren und der Aufruf zur Laufzeit auf die spezifische Implementierung verweist.

````python
class Hund:
    def __init__(self, name, rasse):
        self.name = name
        self.rasse = rasse

    def bellen(self):
        return f"{self.name} sagt: Wuff!" 

class Mops(Hund):
    def bellen(self):
        return f"{self.name} sagt: Wuff! (aber anders)"

def hund_bellen(hund):
    print(hund.bellen())

mein_hund = Hund("Bello", "Labrador")
hunde = [mein_hund, Mops("Fido", "Mops")]
for h in hunde:
    hund_bellen(h)
````

### Objektorientierte Analyse (OOA)
Die für die *Objektorientierte Programmierung* erforderliche Vorgehensweise erfordert vor dem Erstellen von Code eine genaue [*Analyse (OOA)*](https://de.wikipedia.org/wiki/Objektorientierte_Analyse_und_Design) der Gegebenheiten, die Grundlage eines anschließenden *Entwurfs (OOD)* ist.

Die Analyse dient dazu, zu ermitteln,
* welche Klassen benötigt werden, 
* festzulegen wofür sie **exakt** und abgegrenzt von allen anderen Aufgaben zuständig sind,
* welche Daten die Klassen enthalten und 
* welche Methoden sie benötigen um ihren (eindeutigen, klar umrissenen Zweck) zu erfüllen.

#### Objektorientierte Analyse (OOA) Beispiel Supermarktrechnung
```
Bananen              2.00      2.14
Käse                 3.00      3.21
Gurke                0.80      0.86
Gartenbank          50.00     59.50
Buntstifte           2.50      2.97
```
Diese Liste ist ein Beispiel, kann aber auch andere, mehr oder weniger Einzelpositionen haben. Die Rechnung **hat** also beliebig viele Einzelpositionen. Eine **hat**-Beziehung (Sprich: "A hat \<Anzahl\> B") bedeutet in einer objektorientierten Analyse, dass es in der Rechnung eine Objektvariable (ein **Objektattribut**) geben muss, das beliebig viele Objekte vom Typ "Einzelposition" aufnehmen kann - also eine [Liste](https://docs.python.org/3/library/stdtypes.html#typesseq-list).

Dazu ein Nachtrag zu Objektattributen:<br>
Ein Objekt **muss** dafür sorgen, dass seine Attribute **niemals** ungültige, unbrauchbare Werte enthält (sonst kann es nicht funktionieren). Dazu müssen Objektattribute **grundsätzlich** **<u>privat</u>** sein. Änderungen daran dürfen nur über Objektmethoden erfolgen, in denen das Objekt die neuen Werte auf Gültigkeit prüfen kann - und **muss**.<br>
Die Objektdaten sind, da sie privat sind (sein müssen!) vor der "Außenwelt" gekapselt. Der Begriff dazu ist [**Kapselung**](https://de.wikipedia.org/wiki/Datenkapselung_(Programmierung)).
Für Listen bedeutet das: Listen dürfen niemals als solche von einer Objektmethode zurückgegeben werden, da die durch eine Referenz nach außen gereichte Liste beliebig geändert werden kann (Elemente löschen, zufügen, manipulieren).

**Zuständigkeit** der Rechnung: Liste der Einzelpositionen der Rechnung führen/verwalten.

Um die Einzelpositionen der Rechnung zufügen zu können (der Vorgang entspricht dem Scannen der einzelnen Artikel an einer Kasse), benötigt die Rechnung eine Objektmethode: `addLineItem(lineItem)`.
Um auf die Einzelpositionen zugreifen zu können (**ohne Zugriff auf die ganze Liste zu geben, die dann von außen frei manipulierbar wäre**), braucht es zwei Methoden für den indizierten Zugriff:
`getLineItemCount()` für die Ermittlung der Anzahl der vorhandenen Einzelpositionen und 
`getLineItem(idx)`, um eine bestimmte Einzelpostion auslesen zu können.

##### Einzelposition (Rechnungsposition)
Eine Einzelposition (`line item`) wie diese
```
Bananen              2.00      2.14
```
**hat** folgende Eigenschaften (Eigenschaften von Objekten sind Objektattribute):
* eine Bezeichnung,
* einen Nettopreis und
* einen Bruttopreis.
Daraus ergeben sich die (privaten) Objektattribute "Bezeichnung" (`caption`), Nettopreis (`net price`) und Bruttopreis (`gross price`).

Die **Zuständigkeit** der Objekte dieser Klasse ist: Eine Rechnungspostion repräsentieren.

Um die Daten aus dem Objekt lesen zu können, braucht das Objekt Zugriffsmethoden, sog. Getter-Methoden (`getCaption()`,`getNetPrice()`, `getGrossPrice()`).

Um die Bezeichnung, den Nettopreis und den Bruttopreis "in das Einzelposition-Objekt hinein zu bekommen", benötigt die Klasse einen Konstruktor (hier `__init__(self, caption, netPrice, grossPrice)`), der
* diese Daten übernimmt, 
* die privaten Attribute anlegt und 
* mit den übergebenen Daten initialisiert.

Daher werden solche Konstruktoren **Initialisierungskonstruktor** genannt.

##### Spezialisierte Rechnungspositionen
Für Lebensmittel (mit $7\%$ Mehrwertsteuer) und Non-Food-Artikel (mit $19\%$ Mehrwertsteuer) berechnet sich der Bruttopreis direkt aus dem Nettopreis.<br>
Beim Anlegen einer solchen Einzelposition ist somit nur die Bezeichnung und der Nettopreis erforderlich. 

Daher werden $2$ zusätzliche Klasse für Lebensmittel-Einzelpositionen und Nonfood-Einzelpositionen benötigt.

Der Konstruktor beider Klassen 
* übernimmt somit nur den Bezeichner und den Nettopreis eines Artikels.
* Diese werden im Konstruktor dem Konstruktor der Standard-Einzelpositions-Klasse übergeben. Zusätzlich wird 
* der korrekt berechneten Bruttopreis übergeben.

Die **Zuständigkeiten** sind: Lebensmittel-Rechnungsposition und Nonfood-Artikel-Rechnungspostion repräsentieren.

##### Summenbildung der Netto- und Bruttopreise
Bei der Ausgabe der Rechnung werden die Summen der Beträge benötigt:
```
Summen               58.3     68.68
```

Das Summieren könnte (nach funktionsorientierter Denkweise) "irgendwo" im Code gemacht werden - am liebsten da, wo ohnehin die Liste der Einzelpositionen verarbeitet wird - beim Ausdrucken.<br>
Mit Blick auf das [**Single responsibility Principle**](https://de.wikipedia.org/wiki/Single-Responsibility-Prinzip) ist "Ausdrucken" aber eine andere Zuständigkeit als "Aufsummieren".<br>
Und damit muss sich eine andere Klasse um das Summieren kümmern.

Das Ergebnis, die Summenzeile **hat**

* eine Bezeichnung
* einen Nettopreis und
* einen Bruttopreis

Damit sieht sie genau wie eine Einzelpostion aus und kann bei ber Ausgabe genau wie eine solche behandelt werden.

**Zuständigkeit**: Berechnung der Summe von Netto- und Bruttopreisen einer Rechnung.

##### Ausgabe der Rechnung
Die letzte Aufgabe ist, die Rechnung zu drucken, wofür eine weitere Klasse zuständig ist.
**Zuständigkeit**: Ansprechende Druckausgabe einer Rechnung.

Die Druckerklasse 
* sorgt für eine "schicke Ausgabe" durch Kopf-, Fuß- und Trennzeilen, 
* liest die Einzelpositionen der Rechnung 
* lässt die Summen ermitteln und gibt auch diese als Einzelpostion aus.

### Objektorientiertes Design (OOD) Beispiel Supermarktrechnung
Aus der Analyse ergibt sich folgendes Klassendesign, dass zunächst (schriftlich) festhalten wird:

<pre>
Bill                   Rechnung
__init__():            Konstruktor, Erstellen der privaten Attribute, Liste der Einzelpositionen, eine Summe.

LineItem:              Implementierung einer allgemeinen Einzelposition die die erforderlichen Daten einer speichert.

getCaption():          Gibt den Bezeichner zurück.
getNetPrice():         Gibt den Nettopreis zurück.
getGrossPrice():       Gibt den Bruttopreis zurück.

FoodLineItem -> LineItem: Einzelposten für Lebensmittel, 7% Mehrwertsteuer.

NonfoodLineItem -> LineItem: Einzelposten für Lebensmittel, 19% Mehrwertsteuer.

                       Nettopreis sowie den berechneten Nonfood-Artikel an den LineItem - Superkonstruktor weiter.

**Hinweis:**<br>
Auf den ersten `self`-Parameter in den Methoden wurde aus Gründen der Übersichtlichkeit verzichtet

**<u>!!! GANZ DRINGENDE WARNUNG !!!</u>**

Im folgenden Beispiel werden die Währungsbeträge mit einfachen Python-Float-Werten gespeichert und berechnet.<br>
Der **einzige** Grund dafür ist, das Demo-Skript nicht mit Details zu belasten, die den eigentlichen Zweck verdecken.<br>
Motto: "Den Wald vor lauter Bäumen nicht sehen".

Wegen der zwangsläufigen Rundungsfehler von Dezimal-Fließkommaberechnungen mit Binär-Fließkommazahlen darf das in realen Anwendungen **<u>auf keinem Fall</u>** so implementiert werden.<br>
Ein geeigneter Datentyp zur Speicherung und Berechnung von Währungen ist die Klasse [`Decimal`](https://docs.python.org/3/library/decimal.html) aus dem Package `decimal`, die durchgehend für alle Währungsbeträge verwendet werden müsste.<br>

````python
# Loesung 2 Rechnung zum Übungsblatt OOP

# Basisklasse für alle Einzelpositionen eine Supermarkts
# Verwaltet die in jedem Fall nötigen Daten "Bezeichnung" (caption) und "Nettopreis" (netPrice)
class LineItem:

    def __init__(self, caption, netPrice, grossPrice):
        self.__caption__ = caption
        self.__netPrice__ = netPrice
        self.__grossPrice__ = grossPrice

    def getCaption(self):
        return self.__caption__

    def getNetPrice(self):
        return self.__netPrice__

    def getGrossPrice(self):
        return self.__grossPrice__

# Konkrete Einzelposition-Klasse für Lebensmittel mit 7% Mehrwertsteuer
# Berechnet den Bruttopreis mit 7% MWSt aus dem Nettopreis
class FoodLineItem(LineItem):

    def __init__(self, caption, netPrice):
        super().__init__(caption, netPrice, netPrice * 1.07)

# Konkrete Einzelposition-Klasse für Nonfood-Artikel mit 19% Mehrwertsteuer
# Berechnet den Bruttopreis mit 19% MWSt aus dem Nettopreis
class NonfoodLineItem(LineItem):

    def __init__(self, caption, netPrice):
        super().__init__(caption, netPrice, netPrice * 1.19)

# Rechnung
class Bill:
    # Initialisiert die erforderlichen Attribute
    def __init__(self):
        self.__items__ = []

    # Fügt ein Einzelpositionsobjekt in die Liste der Einzelpositionen der Rechnung ein.
    def addLineItem(self, lineItem):
        # Da als 2. Parameter UNBEDINGT ein Einzelpositions-Objekt benötigt wird,
        # (also Typ LineItem) muss der Typ des Parameter geprüft werden.
        if (not isinstance(lineItem, LineItem)):
            raise ValueError(f"Expected Object of type {LineItem}, got {type(lineItem)}")
        self.__items__.append(lineItem)
        
    # Anzahl der Rechnungspostisionen
    def getLineItemCount(self):
        return len(self.__items__)
    
    # Rechnungspostion an der Stelle idx zurück geben.
    def getLineItem(self, idx):
        if (len(self.__items__) == 0):
            raise ValueError("Keine Rechnungspositionen vorhanden")
        if (idx < 0 or idx >= len(self.__items__)):
            raise ValueError(f"Erwarte Index von 0 bis {len(self.__items__)-1}, ist aber {idx}")
        return self.__items__[idx]
    
class BillSummingUnit:
    def sum(self, bill):
        if (not isinstance(bill, Bill)):
            raise ValueError(f"Expected Object of type {Bill}, got {type(bill)}")
        netPriceSum=0
        grossPriceSum=0
        for itemIdx in range(0, bill.getLineItemCount()):
            lineItem = bill.getLineItem(itemIdx)
            netPriceSum += lineItem.getNetPrice()
            grossPriceSum += lineItem.getGrossPrice()
        return LineItem("Summen", netPriceSum, grossPriceSum)
    
# Druckt ein Objekt der Klasse Bill auf STDOUT aus.
class BillPrinter:

    # Private (Hilfs-)Methode zur Ausgabe einer einzelnen Einzelposition.
    def __printItem__(self, item):
       print (f"{item.getCaption():15}{item.getNetPrice():10.2f}{item.getGrossPrice():10.2f}") 
    
    # Die eigentliche Druckfunktion
    def print(self, bill):
        if (not isinstance(bill, Bill)):
            raise ValueError(f"Expected Object of typ {Bill}, got {type(bill)}")
        
        # Kopfzeilen der Rechnung
        print("Rechnung")
        print()
        print(f'{"Artikel":15}{"Netto":>10}{"Brutto":>10}')
        print("=" * 35)

        # Alle Einhelpostionen 
        for itemIdx in range(0, bill.getLineItemCount()):
            lineItem = bill.getLineItem(itemIdx)
            self.__printItem__(lineItem)

        # Optische Trennung Einzelpositionen - Summe
        print("-" * 35)
        
        # Summe berechnen (lassen!!! (nicht zuständig)) und ausgeben
        self.__printItem__(BillSummingUnit().sum(bill))
        
        # Unterstreichung der Summen.
        print(f'{(""):15}{"=" * 8:>10}{"=" * 8:>10}')

#
# Testfunktion
#
# Der Ablauf hier ist der gesamte Programmablauf durch Instanziierung 
# der erforderlichen Objekte sowie Aufruf der erforderlichen Methoden.
#

# Erstellen / öffnen der Rechnung
bill = Bill()

# Einzelpositionen erstellen und der Rechnung zufügen
bill.addLineItem(FoodLineItem("Bananen", 2.0))
bill.addLineItem(FoodLineItem("Käse", 3.0))
bill.addLineItem(FoodLineItem("Gurke", 0.8))
bill.addLineItem(NonfoodLineItem("Gartenbank", 50.0))
bill.addLineItem(NonfoodLineItem("Buntstifte", 2.5))

# Rechnung drucken und dabei die Summe berechnen
BillPrinter().print(bill)
````
