## Erstes Beispiel: Erstellen einer Klasse

### Definition der Klasse

Wir erstellen im folgenden Code Abschnitt eine einfache Klasse `Hund` mit den Eigenschaften (Attributen) `name` und `rassse`, sowie der Methode `bellen()`.

```python
class Hund:
    def __init__(self, name, rasse):
        self.name = name
        self.rasse = rasse

    def bellen(self):
        return f"{self.name} sagt: Wuff!"
```

### Instanziierung / Initialisierung der Klasse:

Nun können wir eine *Instanz* der Klasse (also eine Variable vom Typ `Hund`) anlegen.

```python
mein_hund = Hund("Bello", "Labrador")

print(mein_hund.bellen())  # Ausgabe: Bello sagt: Wuff!

type(mein_hund)  # Ausgabe: <class '__main__.Hund'>
```

### `self`, `__init__()`,  `__new__()` ujnd `__del__()`

In der Definition der Klasse `Hund` mag "*der Bezeichner*" `self` auffallen, sowie die Funktion `__init__()`.

- `self` stellt eine Referenz auf die aktuelle Instanz der Klasse dar. Wenn Sie ein Objekt einer Klasse erstellen, können Sie mit `self` auf dessen Daten und Funktionen zugreifen.
- `self` muss immer als erstes Argument in den Methoden der Klasse stehen. Dies zeigt an, dass die Methode einer Instanz der Klasse aufgerufen wird.
- Mit `self` können Instanzattribute (Variablen) definieren werden und darauf zugegriffen werden. 
- Die Nutzung des Bezeichners `self` ist (nur) eine Konvention. `self` ist kein reserviertes Wort; Sie könnten es theoretisch auch anders benennen. Es ist jedoch ratsam, bei `self` zu bleiben, um den Code verständlich zu halten.

- Der Konstruktor `__init__()` initialisiert die Attribute `name` und `rassse` und verwendet `self`, um sicherzustellen, dass diese Attribute zur Instanz gehören.
- Der Konstruktor ist eine spezielle Methode in Python, die aufgerufen wird, wenn ein neues Objekt der Klasse erstellt wird. Die Methode `__init__()' initiiert die Eigenschaften des Objekts und wird verwendet, um ihre Attribute festzulegen.
- Die `__init__()`-Methode wird automatisch aufgerufen, nachdem das Objekt erfolgreich erstellt wurde.
  - In Beispiel oben wird `__init__()` verwendet, um die Attribute `name` und `rassse` des `mein_hund`-Objekts festzulegen, sobald es instanziiert wird.
- Es gibt neben dem Konstruktor `__init()` noch die Funktion `__new__` die  für die Erstellung des Objekts zuständig ist (während `__init__()` für die Initialisierung der Attribute des Objekts verantwortlich ist. In den meisten Fällen werden Sie nur `__init__()` verwenden, es sei denn, Sie müssen spezifische Kontrolle über den Instanziierungsprozess haben, wobei Sie `__new__()` benötigen könnten.

Die folgende Tabelle fasst die Unterschiede zwischen `__new__()` und `__init__()` zusammen:

Aspekt        | `__new__()`                            | `__init__()`
--------------|----------------------------------------|-----------------------------------------------
Aufruf        | Vor der Erstellung des Objekts         | Nach der Erstellung des Objekts
Zweck         | Erstellt das Objekt und gibt es zurück | Initialisiert das Objekt
Notwendigkeit | Wird selten benötigt.                  | Häufig verwendet, z.B. um Attribute zu setzen

Im folgenden Beispiel wird `__new__()` verwendet um eine Nachricht auszugeben, bevor das neue Objekt erstellt wird. Hier ist zu beachten, dass der Aufruf von `super()` in `__new__()` notwendig ist, um tatsächlich das neue Objekt zu erstellen (dazu später mehr).

Analog zum Konstruktor gbt es noch den Destruktor `__del__()`, der in dem Moment aufgerufen wird, kurz bevor ein Objekt zerstört wird. Der Aufruf des Destruktors macht z.B. Sinn, um geöffnete Dateien zu schließen bevor die entsprechende Instanz der Klasse zerstört wird.

```python
class Auto:
    def __new__(cls, *args, **kwargs):
        print("Ein neues Objekt wird erstellt.")
        # Objekt erstellen
        return super(Auto, cls).__new__(cls)

    def __init__(self, marke, modell):
        self.marke = marke
        self.modell = modell
        print(f"Auto initialisiert: {self.marke} {self.modell}")

    def __del__(self):
        print(f"Auto gelöscht: {self.marke} {self.modell}")

# Objekt erstellen
mein_auto = Auto("Volkswagen", "Golf")

# instanz löschen
del mein_auto
```

### Klassenmethoden

Im vorherigen Beispiel haben wir statt `self`den Bezeichner `cls` verwendet. Dieser deutet auf eine **Klassenmethode** hin. 

Klassenmethoden sind Methoden, die auf die *Klasse selbst* und nicht auf eine *Instanz der Klasse* operieren. Sie werden verwendet, um auf Klassendaten oder zur Erstellung von Instanzen über verschiedene Konstruktor-Logiken hinweg zuzugreifen. Klassenmethoden werden mit dem Decorator `@classmethod` gekennzeichnet.

- Das erste Argument einer Klassenmethode ist immer `cls`, das auf die Klasse selbst verweist (nicht auf eine Instanz). Hiermit hat man Zugriff auf die Klasse.
- Klassenmethoden sind nützlich, um Funktionalitäten zu implementieren, die für die Klasse insgesamt relevant sind und nicht für einzelne Objekte (Instanzen).

```python
class Fahrzeug:
    @classmethod
    def info(cls):
        return f"Dies ist eine Instanz der Klasse {cls.__name__}."

print(Fahrzeug.info())  # Ausgabe: Dies ist eine Instanz der Klasse Fahrzeug.
```

```python
class Buch:
    allgemeine_kategorie = "Literatur"

    def __init__(self, titel, autor):
        self.titel = titel
        self.autor = autor

    @classmethod
    def aus_kategorie(cls):
        return cls.allgemeine_kategorie

# Aufruf der Klassenmethode
print(Buch.aus_kategorie())  # Ausgabe: Literatur
```

In diesem Beispiel ist `aus_kategorie()` eine Klassenmethode, die die allgemeine Kategorie des Buches zurückgibt.
Sie verwendet `cls`, um auf die Klasse Buch zuzugreifen und damit auf das Klassenattribut `allgemeine_kategorie`.

Klassenmethoden können als *alternative Konstruktoren* dienen, um Instanzen auf verschiedene Weisen zu erstellen. Hier ein Beispiel:

```python
class Auto:
    def __init__(self, marke, modell):
        self.marke = marke
        self.modell = modell

    @classmethod
    def von_string(cls, auto_string):
        marke, modell = auto_string.split('-')
        return cls(marke, modell)  # Erstellung einer Instanz

# Alternative Konstruktion
mein_auto = Auto.von_string("Volkswagen-Golf")
print(mein_auto.marke)  # Ausgabe: Volkswagen
print(mein_auto.modell)  # Ausgabe: Golf
```

Die `math`-Bibliothek in Python, die wir schon am Rande kennen gelernt haben, ist ein gutes Beispiel für die Verwendung von Klassenmethoden. Obwohl die `math`-Bibliothek nicht in Klassen aufgebaut ist, können wir dieselben Konzepte auf Klassen mit ähnlichem Verhalten anwenden. 

```python
class Mathe:
    @classmethod
    def quadrat(cls, zahl):
        """Berechnet das Quadrat einer Zahl."""
        return zahl ** 2

    @classmethod
    def umrechnen(cls, von, zu, faktor):
        """Konvertiert von einer Einheit in eine andere unter Verwendung eines Faktors."""
        return von * faktor / zu

# Nutzung der Klassenmethoden
print(Mathe.quadrat(4))  # Ausgabe: 16
print(Mathe.umrechnen(100, 1, 0.001))  # Ausgabe: 0.1 (z.B. (100) mm in (1) m)
```

Diese Klassenmethoden im Beispiel oben ermöglicht es, die Methode ohne Instanziierung der Klasse aufzurufen.

### Klasseneigenschaften

In Fällen, in denen eine Klasse Attribute oder Eigenschaften hat, bietet eine Klassenmethode Zugriff auf diese. Dies ist besonders nützlich für Fabrikmethoden oder zur Verwendung von Klassendaten.

Auf Klasseneigenschaften (hier am Beispiel der Variable `faktor` wird dann mit `cls.faktor` zugegriffen (statt über `self.faktor` (wenn die Variable im Konstruktor definiert worden wäre)).

```python
class Beispiel:
    faktor = 2

    @classmethod
    def multipliziere(cls, zahl):
        return zahl * cls.faktor

print(Beispiel.multipliziere(5))  # Ausgabe: 10
```

### Kapselung
Daten (Attribute) und die dazugehörigen Funktionen (Methoden) werden in einer Klasse in einer Einheit gebündelt und der direkten Zugriff auf die Daten von außen wird beschränkt. Dies dient zum
- Schutz der Datenintegrität und zur 
- Reduzierung der Komplexität,
indem Benutzer nur mit definierten Methoden interagieren.

Die folgende Klasse *implementiert ein Bankkonto*. Auf den Kontostand (Variable `__kontostand`) kann nicht direkt zugegriffen werden (nur über die zur Verfügung gestellten Methoden (z.B. `einzahlen()`, `abheben()`,...))

```python
class Bankkonto:
    def __init__(self, kontostand=0):
        self.__kontostand = kontostand  # private Variable

    def einzahlen(self, betrag):
        self.__kontostand += betrag

    def abheben(self, betrag):
        if betrag <= self.__kontostand:
            self.__kontostand -= betrag
        else:
            print("Nicht genügend Guthaben!")

    def anzeigen(self):
        return self.__kontostand
```

```python
mein_konto = Bankkonto()
print(mein_konto.anzeigen())  # Ausgabe: 0
mein_konto.einzahlen(100)
print(mein_konto.anzeigen())  # Ausgabe: 100
mein_konto.abheben(50)
print(mein_konto.anzeigen())  # Ausgabe: 50
mein_konto.abheben(100)  # Ausgabe: Nicht genügend Guthaben!
```

Auf die Variable `__kontostand` der Klasse `Bankkonto` kann nicht von außerhalb der Klasse zugegriffen werden, da diese *privat* ist (dazu später noch mehr).

```python
print(mein_konto.__kontostand)  # Fehler: AttributeError, da __kontostand privat ist
```

#### Verwendung von Unterstrichen in Python

Die doppelten Unterstriche in Python, wie z.B. bei `__init__`, haben spezielle Bedeutungen und dienen hauptsächlich der Namensverknüpfung (Name Mangling) und der Vermeidung von Namenskonflikten. Hier sind die wichtigsten Punkte:

1. Spezielle Methoden (Dunder-Methoden)
   - Die doppelten Unterstriche markieren spezielle Methoden, die oft als „Dunder-Methoden“ (Double Underscore Methods) bezeichnet werden. Beispielhafte Methoden sind:
     - `__init__`: Der Konstruktor, der beim Erstellen eines Objekts aufgerufen wird.
     - `__str__`: Definiert, wie das Objekt als String dargestellt wird.
     - `__repr__`: Gibt eine offizielle String-Repräsentation des Objekts zurück.
   - Diese Methoden haben spezifische Aufgaben und können von Python intern aufgerufen werden. 
   - Die doppelten Unterstriche sind nicht nur eine Konvention, sondern ein wichtiges Merkmal der Python-Programmierung, das die Struktur und Sicherheit von Code verbessert. Sie führen zu einer klareren Trennung zwischen öffentlichen und privaten Attributen und Methoden.

2. Namensverknüpfung (Name Mangling)
   - Bei Attributen, die mit doppelten Unterstrichen definiert sind (z.B. __attribute), wandelt Python den Namen intern um, um Namenskonflikte in abgeleiteten Klassen zu vermeiden. Das bedeutet, dass:
    - __attribute in der Klasse MyClass intern als _MyClass__attribute gespeichert wird.
Dies macht es schwierig, direkt auf diese Attribute von außen zuzugreifen, fördert aber den sicheren Zugriff auf die Attribute innerhalb der Klasse.

3. Kapselung von Attributen
   - Durch die Verwendung von doppelten Unterstrichen wird eine Form von Kapselung erreicht. Das ermöglicht es, interne Implementierungsdetails eines Objekts zu verbergen.

Beispiel
Hier ist ein einfaches Beispiel, das die Verwendung von `__init__` und doppelten Unterstrichen verdeutlicht:

```python
class BeispielKlasse:
    def __init__(self, wert):
        self.__wert = wert  # private Attribut mit Namensverknüpfung
        self._wert = wert   # geschütztes Attribut
        self.wert = wert    # öffentliches Attribut

    def zeige_wert(self):
        return self.__wert

objekt = BeispielKlasse(10)
print(objekt.zeige_wert())  # Ausgabe: 10
```

### Platzierung von Unterstrichen

- Am Anfang eines Bezeichners:

  - `self._attribut`: wird als **protected** betrachtet, d.h. es soll nur innerhalb der Klasse und von abgeleiteten Klassen verwendet werden.
  - `self.__attribut`: wird durch Namensverknüpfung geschützt und ist **privat**“** für die Klasse.

- Am Ende eines Bezeichners:

  - `name_` oder ``name__: Hierbei handelt es sich normalerweise um die Konvention, um Reservierungskonflikte mit Schlüsselwörtern oder um spezielle Funktionen zu vermeiden. Der Unterstrich am Ende wird verwendet, um Konflikte mit eingebaute Namen zu vermeiden.


```python
print(objekt._wert)  # Dies sollte man nicht tun, aber es funktioniert
print(objekt.wert)   # Öffentliches Attribut, kein Problem
```

```python
print(objekt.__wert)  # Dies führt zu einem Fehler
```

### Vererbung

Bei der Vererbung können untergeordnete Klassen die übergeordneten Klassenmethoden überschreiben oder erweitern und so anpassbare Funktionalität bereitstellen.

Beispiel 1:

```python
class Hund:
    def __init__(self, name, rasse):
        self.name = name
        self.rasse = rasse

    def bellen(self):
        return f"{self.name} sagt: Wuff!"

class Mops(Hund):
    def bellen(self):
        return f"{self.name} sagt: Wuff! (aber anders)"
```

Beispiel 2:

```python
class Tier:
    def sprechen(self):
        return "Tiergeräusch"

class Hund(Tier):  # Hund erbt von Tier
    def sprechen(self):
        return "Wuff!"
```

```python
class Tier:
    @classmethod
    def geräusch(cls):
        return "Ein Tiergeräusch"

class Hund(Tier):
    @classmethod
    def geräusch(cls):
        return "Wuff!"

print(Hund.geräusch())  # Ausgabe: Wuff!
```

### Erben von mehreren Klassen

leichzeitige Vererbung (auch Multiple Vererbung genannt) ist ein Feature in Python, bei dem eine Klasse von mehr als einer Oberklasse erbt. Dies kann nützlich sein, um Funktionen und Eigenschaften aus mehreren Basisklassen in einer Unterklasse zu kombinieren. Im folgenden Code-Abschnitt erbt die Klasse `Auto` von zwei Klassen (`Motorfahrzeug` und `Kraftstoff`).

```python
# Basis-Klasse 1
class Motorfahrzeug:
    def __init__(self, marke, modell):
        self.marke = marke
        self.modell = modell
    
    def fahrzeug_info(self):
        return f"{self.marke} {self.modell}"

# Basis-Klasse 2
class Kraftstoff:
    def __init__(self, kraftstoff_typ):
        self.kraftstoff_typ = kraftstoff_typ
    
    def kraftstoff_info(self):
        return f"Mein Fahrzeug läuft mit {self.kraftstoff_typ}."

# Unterklasse, die von beiden Basisklassen erbt
class Auto(Motorfahrzeug, Kraftstoff):
    def __init__(self, marke, modell, kraftstoff_typ):
        Motorfahrzeug.__init__(self, marke, modell)
        Kraftstoff.__init__(self, kraftstoff_typ)

    def auto_info(self):
        return f"{self.fahrzeug_info()}. {self.kraftstoff_info()}"

# Verwendung
mein_auto = Auto("Volkswagen", "Golf", "Benzin")
print(mein_auto.auto_info())
```

### Polymorphismus

Polymorphismus ermöglicht es, dass verschiedene Klassen Methoden mit dem gleichen Namen implementieren und der Aufruf zur Laufzeit auf die spezifische Implementierung verweist.

```python
class Hund:
    def __init__(self, name, rasse):
        self.name = name
        self.rasse = rasse

    def bellen(self):
        return f"{self.name} sagt: Wuff!" 

class Mops(Hund):
    def bellen(self):
        return f"{self.name} sagt: Wuff! (aber anders)"

def hund_bellen(hund):
    print(hund.bellen())

mein_hund = Hund("Bello", "Labrador")
hunde = [mein_hund, Mops("Fido", "Mops")]
for h in hunde:
    hund_bellen(h)
```

## Abstraktion

Abstraktion bedeutet, die wesentlichen Merkmale einer Entität zu isolieren, während unwichtige Details verborgen bleiben, um die Komplexität zu reduzieren.

- Python verfügt über das Modul `abc` (*Abstract Base Classes*), um abstrakte Klassen zu definieren.
- Abstrakte Klassen können keine Instanzen erzeugen und enthalten mindestens eine abstrakte Methode, die von den Unterklassen implementiert werden muss.

```python
from abc import ABC, abstractmethod

# Abstrakte Klasse
class Fahrzeug(ABC):
    
    @abstractmethod
    def fahren(self):
        pass

# Unterklasse, die die abstrakte Methode implementiert
class Auto(Fahrzeug):
    def fahren(self):
        return "Das Auto fährt."

class Fahrrad(Fahrzeug):
    def fahren(self):
        return "Das Fahrrad fährt."

# Verwendung
def fahrzeug_fahren(fahrzeug: Fahrzeug):
    print(fahrzeug.fahren())

meine_auto = Auto()
mein_fahrrad = Fahrrad()

fahrzeug_fahren(meine_auto)  # Ausgabe: Das Auto fährt.
fahrzeug_fahren(mein_fahrrad)  # Ausgabe: Das Fahrrad fährt.
```
