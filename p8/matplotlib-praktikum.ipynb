Erzeugen Sie nun $2$ Sinussignale 

\begin{align}
x_1(t) &= \sin(2 \pi f_1 t) \text{  und} \\
x_2(t) &= \sin(2 \pi f_2 t) 
\end{align}

mit den Frequenzen $f_1=60$ Hz und $f_2=200$ Hz (jeweils der Länge $0.5\,\mathrm{sec}$) und addieren Sie diese. Stellen Sie die Signale sowie die Addition der Signale unter Verwendung des [`plt.subplot()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplot.html) Befehls dar.

Probieren Sie dies aus.

Was hören Sie?
