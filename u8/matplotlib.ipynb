## Matplotlib importieren

So wie wir die Kurzform `np` für NumPy (oder `pd` für Pandas) verwenden, nutzen wir auch für Matplotlib einige Standard-Kurzformen:

```python
import matplotlib.pyplot as plt  # for plotting
import numpy as np  # for numerical operations
```

Mit NumPy erzeugte Sinus- oder Kosinussignale ([`np.sin()`](https://numpy.org/doc/stable/reference/generated/numpy.sin.html) oder [`np.cos()`](https://numpy.org/doc/stable/reference/generated/numpy.cos.html)) für eine mit [`np.arange()`](https://numpy.org/doc/stable/reference/generated/numpy.arange.html) oder dem äquivalenten Befehl [`np.linspace()`](https://numpy.org/doc/stable/reference/generated/numpy.linspace.html) erzeugte Laufzeitvariable $x$

$$ 
f(x) = \sin(x).
$$

können wie folgt dargestellt werden:

```python
x = np.linspace(0, 10, 100)

fig = plt.figure()
plt.plot(x, np.sin(x), '-')
plt.plot(x, np.cos(x), '--');
```
