## Aufgaben 2 bis 5 - Sinuswelle erzeugen

Die folgende Funktion `get_sine_wave()` erzeugt ein Sinussignal mit einer bestimmten Frequenz $f$.

Es sei angemerkt, dass es gängige Praxis ist, Funktionen sowie deren Eingabe- und Ausgabeparameter zu beschreiben, wie hier beispielhaft in einem mehrzeiligen Kommentar gezeigt.


````python
def get_sine_wave(frequency_Hz, length_s=1, fs_Hz=8000):
    """
    Return a sine wave with given parameters.
    
    Parameters
    ----------
    frequency_Hz : float
        frequency $f$ of the sine to be generated 
    length_s : float, optional
        length of the sine signal to be generated, default: 1 sec.
    fs_Hz : float, optional
        sampling frequency $f_s$, default: 8000 Hz

    Returns
    -------
    signal
        generated sine signal
    """
    # create time vector $t$
    time_vector = np.linspace(0, length_s,
                              int(length_s * fs_Hz),
                              endpoint=False)
    # return sine of frequency $f$, i.e. variable frequency_Hz
    return np.sin(2 * np.pi * frequency_Hz * time_vector)
````

Der folgende Code erzeugt und visualisiert eine Sinuswelle mit $f=440$ Hz, dem sogenannten [Kammerton](https://en.wikipedia.org/wiki/Concert_pitch "Open Wikipedia for more information on Concert Pitch") bzw. der musikalischen Note **A** auf einer Klaviatur.


````python
# crate a sine of f=440 Hz at a sampling frequency f_s = 8 kHz of 2 seconds length
concert_pitch = get_sine_wave(440, 2, 8000)
# Look at the first 30 ms (240 / 8000 = 0.03 s) of the generated wave
plt.plot(concert_pitch[:240])
plt.xlabel('sample $k$')
plt.ylabel('sine of 440 Hz')
plt.title('First 240 samples of sine of 440 Hz')
None # This command does nothing but prevents printing the result of previous function call
````

Wir können das Signal mit dem Befehl `IPython.display.Audio()` anhören.


````python
# listen to the sound file 
from IPython import display as ipd
ipd.Audio(concert_pitch, rate=8000)
````

Wir können Signale verketten, um ein längeres Signal zu erzeugen.
Als Beispiel wollen wir ein neues Signal mit halber Kammerton-Frequenz erstellen und es dann zweimal abwechselnd mit dem Kammerton wiedergeben.


````python
half_concert_pitch = get_sine_wave(220, 2, 8000)
# Look at the first 0.02 s of the signals

plt.plot(half_concert_pitch[:160], "red",label="sine with $f=200$ Hz")
plt.plot(concert_pitch[:160], "green",label="sine with $f=400$ Hz")
plt.xlabel('samples $k$')
plt.legend(loc='upper right') # create a legend showing the line labels 

alternation = np.concatenate((half_concert_pitch, concert_pitch,
                              half_concert_pitch, concert_pitch))
ipd.Audio(alternation, rate=8000)
````

Wir können beobachten, dass der grüne `concert_pitch` eine Frequenz hat, die genau doppelt so hoch ist wie die des roten Signals `half_concert_pitch`, d. h. zwei Perioden der grünen Kurve passen in eine Periode der roten Kurve (siehe z. B. zwischen den Samples $k=0$ und $k \approx 34$).

Wenn man zwei Signale addiert, entsteht ein Signal, in dem beide Frequenzen hörbar sind.
Das resultierende Signal sieht nicht mehr wie eine reine Sinuswelle aus, aber seine Periodizität ist weiterhin klar erkennbar. 

Die Periodizität ist so deutlich, weil eine Frequenz ein Vielfaches der anderen ist.
Ist dies nicht der Fall, kann es wesentlich länger dauern, bis das Signal sein Muster wiederholt.

````python
plt.figure(figsize=(8,5)) # create figure of size 8 x 5 inches
plt.plot(half_concert_pitch[:160], "red", alpha=0.3, label='1/2 concert pitch ($f=220$ Hz)')
plt.plot(concert_pitch[:160], "green", alpha=0.3, label='concert pitch ($f=440$ Hz)')
plt.plot(half_concert_pitch[:160] + concert_pitch[:160], label='superposition')
plt.legend(loc='upper right')
ipd.Audio(half_concert_pitch + concert_pitch, rate=8000)
````

Die Addition von zwei Sinussignalen mit nahe beieinanderliegenden Frequenzen sieht und klingt folgendermaßen aus:


````python
signal_1 = get_sine_wave(440, 2, 8000)
signal_2 = get_sine_wave(400, 2, 8000)
samples = 600                           # samples used for plotting

plt.figure(figsize=(8,5)) # create figure of size 8 x 5 inches
plt.plot(signal_1[:samples], "red", alpha=0.3, label='sine with $f=440$ Hz')
plt.plot(signal_2[:samples], "green", alpha=0.3, label='sine with $f=400$ Hz')
plt.plot(signal_1[:samples] + signal_2[:samples], label='superposition')
plt.legend()

ipd.Audio(signal_1 + signal_2, rate=8000)  # play sum of two (sine) signals
````

## Aufgabe 6 - Gedämpfte Sinuswelle 

Eine gedämpfte Sinuswelle kann definiert werden als

$$
x(t) = \sin(2 \pi f t) \, \mathrm{e}^{-\alpha t}
$$


Der folgende Code erzeugt ein Sinussignal und eine abklingende Hüllkurve.
Die gedämpfte Sinuswelle wird erzeugt, indem die beiden Signale **elementweise** multipliziert werden.


````python
fs=8000
t=np.arange(0,1,1/fs) # crete time vector of 1 second length
f = 2.5               # 2.5 Hz 
alpha = 1             # daming factor (play around with it and see what happens)

sin  = np.sin(2*np.pi*f*t)
damp = np.exp(-alpha*t)
x = sin * damp

plt.subplot(2,1,1)
plt.plot(t,sin)
plt.ylabel('sin($t$)')

plt.subplot(2,1,2)
plt.plot(t,x, label=r'sin$(2 \pi \cdot ' + str(f) + r' \mathrm{Hz} \cdot t) \cdot \mathrm{e}^{- a t}$')
plt.plot(t,damp, '--', label=r'$\mathrm{e}^{- a t}$')
plt.legend()
plt.xlabel('$t$ in seconds'); # the ; also suppresses output, try to delete it
````

Im Folgenden werden wir die abwechselnden Töne zunächst **ohne Dämpfung** wiedergeben.


````python
concert_pitch_1sec = get_sine_wave(440, 1, 8000)
half_concert_pitch_1sec = get_sine_wave(220, 1, 8000)

alternation_1sec = np.concatenate((half_concert_pitch_1sec, concert_pitch_1sec,
                                     half_concert_pitch_1sec, concert_pitch_1sec))

#plot the tune
plt.plot(alternation_1sec) # you can't see much here, it will get clear together with the next plot

ipd.Audio(alternation_1sec, rate=8000)
````

Unten können wir den Effekt der Dämpfung auf musikalische Töne beobachten – die gedämpfte synthetisierte Melodie klingt natürlicher.
(Diese Technik wird bei der MIDI-Instrumentensynthese verwendet.)


````python
alternation_damped = np.concatenate((half_concert_pitch_1sec* damp, concert_pitch_1sec* damp,
                                     half_concert_pitch_1sec* damp, concert_pitch_1sec* damp))

#plot the tune
plt.plot(alternation_damped)

ipd.Audio(alternation_damped, rate=8000)
````

Eine einfache Zusatzfunktion, die wir benötigen, um ein komplexeres Lied zu erzeugen, ist eine Funktion zur Erzeugung von Stille.
Die folgende Funktion `get_silence()` erzeugt ein Array von $0$-Werten der gewünschten Länge (in Sekunden).


````python
def get_silence(length_s, sample_rate_hz):
    """Return silence for the given length at the given sample rate."""
    return np.zeros(int(length_s * sample_rate_hz))
````

Nun können wir eine Funktion definieren, die entweder die gewünschte Note oder eine Pause erzeugt, basierend auf den oben beschriebenen `left_hand_notes` und `right_hand_notes`.

Beachte, dass die Frequenz des gewünschten Signals berechnet wird durch

$$
f = 440 \cdot 2^{(\text{note-position} / 12) + (\text{octave} - 4)}
$$


Diese Berechnung wird in der untenstehenden Funktion verwendet.


````python
def create_tone(note, duration):
    
    tempo = 5

    if note == "Pause":
        return get_silence(length_s=duration*tempo, sample_rate_hz=8000)

    note_position = {
        "C": -9,
        "C#": -8,
        "D": -7,
        "D#": -6,
        "E": -5,
        "F": -4,
        "F#": -3,
        "G": -2,
        "G#": -1,
        "A": 0,
        "A#": 1,
        "B": 2,
    }

    octave = int(note[-1])
    key = note[:-1]

    frequency_hz = 440 * 2 ** ((note_position[key] / 12) + (octave - 4))

    return get_sine_wave(frequency_Hz=frequency_hz, length_s=duration*tempo, fs_Hz=8000)
````

````python
minlength= min(len(left_hand), len(right_hand)) # cut to same length if necessary
song = left_hand[:minlength]+ right_hand[:minlength]
ipd.Audio(song, rate=8000)
````

Wir verwenden die Bibliothek `librosa`, um die Abtastrate $f_s$ zu ändern. Der folgende Code reduziert das Audiosignal zunächst auf $f_s = 8$ kHz. Dadurch gehen alle Informationen oberhalb von $f_{\mathrm{max}} = 4$ kHz, also die Hälfte der Abtastrate, verloren. Selbst wenn wir das Signal anschließend wieder auf die ursprüngliche Abtastrate von $f_s = 44{,}1$ kHz hochsamplen, gehen die verlorenen Informationen nicht zurück.


````python
import librosa         # we will use the library librosa here for resampling

#donwsample from 44.1 kHz to 8 kHz
mus_8k = librosa.resample(mus_44k, orig_sr=fs44, target_sr=8000); # resample to 8 kHz
#upsample from 8 kHz to 44.1 kHz
mus_44k_2 = librosa.resample(mus_8k, orig_sr=8000, target_sr=fs44); # resample to 44.1 kHz

# visualise
fig=plt.figure(figsize=(12,6)) # create a figure of size 12 x 6 inches

plt.subplot(1,3,1)
plt.specgram(mus_44k, Fs=fs44);
plt.title('original')
plt.xlabel('time $t$')
plt.ylabel('frequency $f$')

plt.colorbar(label='dB');
plt.clim(-150,0)

plt.subplot(1,3,2)
plt.specgram(mus_8k, Fs=8000);
plt.title('downsampled to $8000$ Hz')
plt.xlabel('time $t$')
plt.ylabel('frequency $f$')

plt.colorbar(label='dB');
plt.clim(-150,0)

plt.subplot(1,3,3)
plt.specgram(mus_44k_2, Fs=fs44);
plt.title('after down-/upsampling')
plt.xlabel('time $t$')
plt.ylabel('frequency $f$')

plt.colorbar(label='dB');
plt.clim(-150,0)

plt.tight_layout() # this command ensures that labelsdon't overlap
````

Die drei Zeit-Frequenz-Darstellungen ([Spektrogramme](https://en.wikipedia.org/wiki/Spectrogram)) oben veranschaulichen die Effekte von Down- und Upsampling.

* Das linke Panel zeigt das Spektrogramm der ursprünglichen WAVE-Datei mit $f_s = 44{,}100 ,\mathrm{Hz}$. Man erkennt, dass Energie im Signal bis zu Frequenzen von $\frac{1}{2} f_s = 22{,}050 ,\mathrm{Hz}$ vorhanden ist (wie durch das [Nyquist-Shannon-Abtasttheorem](https://en.wikipedia.org/wiki/Nyquist%E2%80%93Shannon_sampling_theorem) definiert: $f_{\mathrm{max}} = \frac{1}{2} f_s$).

* Das mittlere Panel zeigt das Signal nach Reduktion der Abtastrate auf $f_s = 8{,}000 ,\mathrm{Hz}$. Somit sind nur noch Signalanteile bis $f_{\mathrm{max}} = 4{,}000 ,\mathrm{Hz}$ enthalten (siehe unterschiedliche $y$-Achse).

* Nach erneutem Hochsamplen auf die ursprüngliche Abtastrate $f_s = 44{,}100 ,\mathrm{Hz}$ (rechtes Panel) erkennt man, dass Signalanteile oberhalb von $f = 4{,}000 ,\mathrm{Hz}$ **nicht wiederhergestellt** werden; sie gehen im Down-/Upsampling-Prozess verloren.

## Aufgabe 11 - Blockverarbeitung 

Um die Blockverarbeitung zu veranschaulichen, schneiden wir zunächst ein Stück von $4096$ Abtastwerten aus, beginnend bei $10$ Sekunden, d. h. ab Abtastwert Nr. $80{,}000$ bei einer Abtastrate von $f_s = 8000 ,\mathrm{Hz}$ aus $s[k]$, das im oberen Panel der Abbildung als roter Ausschnitt dargestellt ist und im unteren Panel wiederholt wird.


````python
# lets cut out a piece of the data and cisualise it
start_sample = int(10*fs);                   # start at 10 sec
#print(start_sample)
no_of_samples = 4096;                        # no of samples we want to cut out    
end_sample   = start_sample + no_of_samples; # last sample to be cut out
sample_vec = np.linspace(start_sample, end_sample, no_of_samples)
x1=s[start_sample:end_sample];

plt.figure(figsize=(8,5))
plt.subplot(2,1,1)
plt.plot(np.arange(0,len(s)),s)
plt.ylabel('$x_1[k]$')
plt.plot(sample_vec,x1,'r')
plt.subplot(2,1,2)
plt.plot(sample_vec,x1,'r')
plt.xlabel('$k$')
plt.ylabel('$x_1$[' + str(start_sample) + '...' + str(end_sample) + ']')
plt.title('$x_1[k]$ for ' + str(len(x1)) + ' samples between ' 
          + str(start_sample) + ' to ' + str(end_sample) + 
          ' ($f_s$=' + str(fs) +')')
plt.tight_layout() # this allowes for some space for the title text.
````

````python
# listen to the short cut-out
ipd.Audio(x1, rate=fs)
````

Wie bei Sprachverarbeitungsalgorithmen üblich, wird eine Blockverarbeitung (Sliding-Window-Ansatz) durchgeführt.


````python
# lets cut out a piece of the data
Lw   = 512                    # frame length
Lov  = 1                      # frame overlap factor 
Lhop = int(np.round(Lw/Lov)); # frame hop size

# creating grid of axes for subplots
plt.figure(figsize=(12, 6))
ax1 = plt.subplot2grid(shape=(2, 5), loc=(0, 0), colspan=5)
ax2 = plt.subplot2grid(shape=(2, 5), loc=(1, 0), colspan=1)
ax3 = plt.subplot2grid(shape=(2, 5), loc=(1, 1), colspan=1)
ax4 = plt.subplot2grid(shape=(2, 5), loc=(1, 2), colspan=1)
ax5 = plt.subplot2grid(shape=(2, 5), loc=(1, 3), colspan=1)
ax6 = plt.subplot2grid(shape=(2, 5), loc=(1, 4), colspan=1)
ax_blocks = [ax2, ax3, ax4, ax5, ax6]

# plot signal part in upper panel (axis ax1)
ax1.plot(sample_vec,x1,'r') #
ax1.set_xlabel('$k$')
ax1.set_ylabel('$x_1$[' + str(start_sample) + '...' + str(end_sample) + ']');
ax1.set_title('A piece between sample ' + str(start_sample) + 
          ' and ' + str(end_sample) + ' (of length ' + str(len(x1)) + 
          ') from the 1st channel ($f_s$=' + str(fs) +')')
          
# plot single blocks in lower panels 
clrs = ['g','y','m','b','c','k']; # define a vector of colours
for ii,k in enumerate(range(start_sample,start_sample+5*Lhop,Lhop)):
    block_k_vec = np.arange(k,k+Lw)
    block_sig_vec = x1[ii*Lhop:ii*Lhop+Lw]
    ax1.plot(block_k_vec,block_sig_vec,clrs[ii])

    ax_blocks[ii].plot(block_k_vec,block_sig_vec,clrs[ii])
    ax_blocks[ii].set_xlabel('$k$')
    ax_blocks[ii].set_ylim(-0.35, 0.35)
    ax_blocks[ii].set_title('Block ' + str(ii))

# automatically adjust padding horizontally 
# as well as vertically.
plt.tight_layout()
````

Die obige Abbildung zeigt einige der Frames/Blöcke der Länge $L_{\mathrm{Bl}}=512$ in den unteren Panels. Bei einer Abtastfrequenz von $f_s=8000 ,\mathrm{Hz}$ können wir berechnen, dass die Blocklänge beträgt:

$$
L_{\mathrm{Bl}} = \frac{512 \text{ Samples}}{8000 \,\mathrm{Hz}} = 64 \,\mathrm{ms}
$$



Da bekannt ist, dass ein Sprachsignal im Zeitfenster von ca. $10 .. 30 ,\mathrm{ms}$ (kurzfristig) stationär ist, kann man schließen, dass es wahrscheinlich sinnvoll wäre, kürzere Frames, also ein kleineres $L_{\mathrm{Bl}}$, zu verwenden. Dies wird auch bei den letzten beiden Frames (blaue Farben) deutlich, bei denen die Änderung der Statistik sogar im Zeitbereichssignal beobachtet werden kann (z. B. sieht die erste Hälfte von Block 3 deutlich anders aus als die zweite Hälfte).
